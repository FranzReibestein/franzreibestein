```python
#Find and print the mean of each column.
print(np.mean(matrix, axis=0))
```

```python
#Find and print the overall mean of the array.
print(np.mean(matrix))
```

```python
#Find the max value in each row and the min value in each column.
print(np.min(matrix, axis=0))
print(np.max(matrix, axis=1))
```

```python
flat = matrix.flatten()
print(flat)
print(flat.shape)
```

```python
# the matrix was transformed into a row vector with 12 entries
```

```python
#extract the indices of the min and max values in the flat array
min_id = np.argmin(flat)
max_id = np.argmax(flat)
print(np.argmin(flat))
print(np.argmax(flat))
```

```python
#syntax: np.unravel_index() takes two arguments: first argument is the index of a n element in a flat (1D) array, second argument is the shape of a 2d array.
#so, np.unravel_index() assumes that the 1D array has as many elements as the second argument 2D array. So, it returns the coordinates of of the min or max element of the !D array in the 2D array
coordinates_min = np.unravel_index(min_id, matrix.shape)
coordinates_max = np.unravel_index(max_id, matrix.shape)
print(coordinates_min)
print(coordinates_max)
```

```python
print(f"The coordinates of the min value in the matrix are row {coordinates_min[0]+1}, column: {coordinates_min[1]+1}.") #add 1 because the index starts at 0 but there is no row/column 0.
print(f"The coordinates of the max value in the matrix are row {coordinates_max[0]+1}, column: {coordinates_max[1]+1}.")
```

# 4 Prime star systems

```python
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])
```

```python
#Write a function that returns only the rows where at least one value is a prime number.

#First, I write a function that checks whether a function is a prime number
def prime_number(n):
    if n<2: #natural numbers smaller than 2 are no primes
        return False
    else:
        for i in range(2, n-1): #loop over all numbers from 2 to the given number -1, since 1 and the number divide the given number
            if n%i == 0: #if any of the numbers between 2 and n-1 divides n, n is not a prime number
                return False
            else: 
                return True
def prime_rows(matrix):
    m = np.array(matrix)
    result= [] # initialise an empty list
    for i in range(m.shape[0]): # loop over the row indices of the matrix
        for j in range(m.shape[1]): #loop over the columns in the row
            if prime_number(m[i,j])==True: #in an element is a prime
                result.append(m[i]) #append the whole row to the empty list. technically, the result list, contains several 1D lists of equal length
                break #stop the loop to avoid appending the same row several times when there are several primes in one row
    return np.array(result) #np. array turns the result list into a matrix

arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])

print(prime_rows(arr))

            
```

# 5.2
For the odd-numbered rows, create an alternating pattern starting with 1.

```python
def odd_rows(arr):
    arr1 = arr.copy()
    for i in range(arr1.shape[0]): #loop over teh columns: recall arr1.shape gives a tuple: (#of rows, #of columns)
        if i%2 ==0: # if the row index is even the row number is odd!
            for j in range(arr1.shape[1]): #loop over the columns in the odd rows
                    if (i+j)%2 == 0: arr1[i][j]=1 #if the sum of the indices of the current row and column are even, set this cell to one
    return arr1

h = odd_rows(h)

print(h)
```

# 5.3
Finish the even-numbered rows: starting with a zero, alternating with 1s

```python
def even_rows(arr):
    arr1 = arr.copy()
    for i in range(arr1.shape[0]):
        if i%2 != 0:
            for j in range(arr1.shape[1]):
                    if (i+j)%2 == 0: arr1[i][j]=1
    return arr1

h = even_rows(h)

print(h)
```

# 6. The expanding universe

```python
#You’ve been inspired by a lecture on the expanding universe. Write a function that takes a string
#and a number, and returns that string with the specified number of spaces between each letter.
universe = np.array(["galaxy", "2clusters"])

def expansion(words, number):
    c = []  #defining an empty list
    for i in words: #iterate over the elements in words
        word_str = "" #define an empty string
        for char in i: #iterate over the letters of the element
            word_str += char #add every letter to the word string
            for j in range(number): #add as many spaces after each letter as specified
                word_str += " "
        c.append(word_str.rstrip()) #word_str contains now one word with the required number of strings between the letters. But to the right of the last lesster,
        #the function also inserted strings. rstrip() deletes all the spaces to the right of the last letter of the word_string
    return c

expansion(universe, 1)

    
```

```python
#7.second brightest star
def second_brightest(matrix):
    m = np.array(matrix)
    vector = np.zeros(m.shape[0]) #create a numpy vector, filled with zeros, as many entries as the matrix has columns
    for i in range(m.shape[0]): #loop over the rows of the matrix
        sorted_row=np.sort(m[i]) #sort the row from smallest to biggest
        vector[i] = sorted_row[-2] #put the 2nd largest element (the penultimate element of the sorted rows) into the vector
    return vector

stars = np.random.randint(500, 2000, (5, 5))
print(second_brightest(stars))
```
