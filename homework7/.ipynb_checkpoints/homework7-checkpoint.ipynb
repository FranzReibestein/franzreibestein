10. For the following print statements, please round your answer to two decimal places:
• Print the state closest to the median latitude and by how much.
• Print the state farthest from the median latitude and by how much.
• Print the state closest to the median longitude and by how much.
• Print the state farthest from the median longitude and by how much.
Hint: You can round to two decimal places by attaching :.2f after your variables in f-strings.
Example: print(f"{variable:.2f}")

```python
#closest to median latitude
min_lat_dist = df["Distance to median latitude"].abs().idxmin()  # i need the state with the smallest absolute distance
```

```python
print(f"the state closest to the median latitude is {df["name"][min_lat_dist]} with a distance of {df["Distance to median latitude"][min_lat_dist]:.2f}.")
```

```python
#farthest away from median latitude
max_lat_dist = df["Distance to median latitude"].abs().idxmax()
print(f"The state farthest from the median latitude is {df["name"][max_lat_dist]} with a distance of {df["Distance to median latitude"][max_lat_dist]:.2f}.")
```

```python
#• Print the state closest to the median longitude and by how much.
#• Print the state farthest from the median longitude and by how much.
max_lon_dist = df["Distance to median longitude"].abs().idxmax()
print(f"The state farthest from the median longitude is {df["name"][max_lon_dist]} with a distance of {df["Distance to median longitude"][max_lon_dist]:.2f}.")

min_lon_dist = df["Distance to median longitude"].abs().idxmin()
print(f"The state closest to the median longitude is {df["name"][min_lon_dist]} with a distance of {df["Distance to median longitude"][min_lon_dist]:.2f}.")

```

```python
df_sns = sns.load_dataset("penguins")
df_cl = df_sns.dropna(axis=0)
print(list(df_cl.columns))
#in the previous exercise, there has been a mistake, it doesnt make any sense to delte column with NaN values, but the rows with NaN values
```

```python
#we have these variables
# - species (Adelie, Chinstrap, or Gentoo).
# - island
# - bill_length_mm: Length of the penguin's bill in millimeters.
# - bill_depth_mm: Depth (vertical height) of the penguin's bill in millimeters.
# - flipper_length_mm: Length of the penguin's flipper in millimeters.
# - body_mass_g: Body mass in grams.
# - sex
```

```python
fig, axes = plt.subplots(2,2)
# Top-left: Scatterplot of bill_length_mm vs. bill_depth_mm
sns.scatterplot(
    data=df_cl,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species", #assigning different colors to different species datapoints
    palette="Set1", #farb scheme for the species
    ax=axes[0, 0],
)
axes[0, 0].set_title("Bill Length vs. Bill Depth")
axes[0, 0].set_xlabel("Bill Length (mm)")
axes[0, 0].set_ylabel("Bill Depth (mm)")
axes[0, 0].legend(title="Species")

# Top-right subplot: Scatterplot of two completely different columns (flipper_length_mm vs. body_mass_g)
sns.scatterplot(
    data=df_cl,
    x="flipper_length_mm",
    y="body_mass_g",
    color="teal",
    ax=axes[0, 1],
)
axes[0, 1].set_title("Flipper Length vs. Body Mass")
axes[0, 1].set_xlabel("Flipper Length (mm)")
axes[0, 1].set_ylabel("Body Mass (g)")

# Bottom-left subplot: Histogram of body_mass_g
sns.histplot(
    data=df_cl, x="body_mass_g", bins=20, color="coral", ax=axes[1, 0]
)
axes[1, 0].set_title("Distribution of Penguin Body Mass")
axes[1, 0].set_xlabel("Body Mass (g)")
axes[1, 0].set_ylabel("Count (Number of Penguins)")

# Bottom-right subplot: Categorical boxplot of island vs. flipper_length_mm
sns.boxplot(
    data=df_cl,
    x="island",
    y="flipper_length_mm",
    ax=axes[1, 1],
)
axes[1, 1].set_title("Flipper Length Across Islands")
axes[1, 1].set_xlabel("Island")
axes[1, 1].set_ylabel("Flipper Length (mm)")
plt.tight_layout()
plt.show()
```

Explain what data each of your subplots represents and why you decided on those
columns.
• What did you learn about your dataset from each subplot?

- upper left: plotts bill depth in dependence of bill length across the three species. I learned that Gentto penguins have middle-long, but shallows bills, whereas the Adelie species has rather short, but deep bills.

- upper right: I plotted the body mass in grams in depende of the flipper length in mm. There is a strong positive correlation: Penguins with logner flippers are heavier on average.

- bottom-left: Displays the overall distribution of body mass across all penguins, showing a peak around 3,500–4,000 g and a right tail extending above 5,000 g.

  bottom-right (boxplot): Compares flipper lengths across the three islands. Penguins on Biscoe Island have noticeably longer median flippers than those on Dream or Torgersen.
