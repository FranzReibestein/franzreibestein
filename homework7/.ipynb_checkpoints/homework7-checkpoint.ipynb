6. Compute the median latitude and median longitude of the entire dataset. What are the
values

```python
med_lat = df["latitude"].median()
print(med_lat)
```

```python
med_lon=df["longitude"].median()
print(med_lon)
```

Compute the absolute value distance to the median latitude for each state. Make a new
column in your DataFrame with these values

```python
df["Distance to median latitude"] = df["latitude"]- med_lat
```

8. Compute the absolute value distance to the median longitude for each state. Make a
new column in your DataFrame with these values.

```python
df["Distance to median longitude"] = df["longitude"]- med_lon
```

10. For the following print statements, please round your answer to two decimal places:
• Print the state closest to the median latitude and by how much.
• Print the state farthest from the median latitude and by how much.
• Print the state closest to the median longitude and by how much.
• Print the state farthest from the median longitude and by how much.
Hint: You can round to two decimal places by attaching :.2f after your variables in f-strings.
Example: print(f"{variable:.2f}")

```python
#closest to median latitude
min_lat_dist = df["Distance to median latitude"].abs().idxmin()  # i need the state with the smallest absolute distance
```

```python
print(f"the state closest to the median latitude is {df["name"][min_lat_dist]} with a distance of {df["Distance to median latitude"][min_lat_dist]:.2f}.")
```

```python
#farthest away from median latitude
max_lat_dist = df["Distance to median latitude"].abs().idxmax()
print(f"The state farthest from the median latitude is {df["name"][max_lat_dist]} with a distance of {df["Distance to median latitude"][max_lat_dist]:.2f}.")
```

```python
#• Print the state closest to the median longitude and by how much.
#• Print the state farthest from the median longitude and by how much.
max_lon_dist = df["Distance to median longitude"].abs().idxmax()
print(f"The state farthest from the median longitude is {df["name"][max_lon_dist]} with a distance of {df["Distance to median longitude"][max_lon_dist]:.2f}.")

min_lon_dist = df["Distance to median longitude"].abs().idxmin()
print(f"The state closest to the median longitude is {df["name"][min_lon_dist]} with a distance of {df["Distance to median longitude"][min_lon_dist]:.2f}.")

```
