```python
import numpy as np
import matplotlib.pyplot as plt
```

# 2 Matplotlib

```python
# 1. Write a function called make_sine_wave(x, A, w) that returns: y = A·sin(ωx). Where x is
#an array of input values, A is the amplitude, and ω is the angular frequency.

def make_sine_wave(x, A, w):
    y = np.sin(w*x)*A
    return y

```

```python
#2.
x = np.linspace(0, 2*np.pi, 1000)
```

```python
#3. 
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
# Call your make_sine_wave() function for each subplot.
#• First subplot: Vary only the amplitude
#• Second subplot: Vary only the frequency
#• Third subplot: Vary the amplitude and frequency together.
#Hint: When you aren’t varying a variable, just set it to 1.

```

```python
#4. Create three side-by-side subplots figsize=(25, 5).
fig, axes = plt.subplots(1,3, figsize = (25,5))
for A, w in zip(amplitude, frequency): #looping over the aplitude and frequency arrays at the same time
    # 1. Subplot: varying the amplitude (w = 1)
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f"A = {A}, ω = 1") #in sine wave function, only a varies, w is fixed at 1

    # 2. Subplot: only varying the frequency (A = 1)
    axes[1].plot(x, make_sine_wave(x, 1, w),  label=f"A = 1, ω = {w}")

    # 3. Subplot: varying the frequency and the amplitude
    axes[2].plot(x, make_sine_wave(x, A, w),  label=f"A = {A}, ω = {w}")

titles = ["Varying the amplitude", "Varying the frequency", "Varying amplitude and frequency"]
for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize=18)
    ax.set_xlabel("x (rad)", fontsize=14)
    ax.set_ylabel("y = A · sin(ωx)", fontsize=14)
    ax.legend(fontsize=11, loc="upper right")

plt.show()
```
