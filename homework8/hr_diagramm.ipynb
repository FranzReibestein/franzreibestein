```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("C:/Users/franz/Astro98/course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
print(data.shape)
```

```python
#2642 rows, 2 columns
```

```python
magnitude = data[:,0]
```

```python
temperature = data[:,1]
```

Print the minimum and maximum value of the magnitude variable. What stars do they
correspond to?

```python
#row index corresponds to a star
idx_min_mag = np.argmin(magnitude)
idx_max_mag = np.argmax(magnitude)
idx_min_temp = np.argmin(temperature)
idx_max_temp = np.argmax(temperature)
```

```python
print(f"Temperature min: {temperature[idx_min_temp]}; corresponds to Star in row: {idx_min_temp}")
print(f"Temperature max: {temperature[idx_max_temp]}; corresponds to Star in row: {idx_max_temp}")
print(f"Magnitude min: {magnitude[idx_min_mag]};; corresponds to the brightest star to be found in row: {idx_min_mag}")
print(f"Magnitude max: {magnitude[idx_max_mag]}; corresponds to the dimmest star to be found in row: {idx_max_mag}")
```

```python
#3.3
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(log_magnitude, log_temperature, label = "stars")
plt.gca().invert_yaxis()
plt.gca().invert_xaxis()
plt.xlabel("log Magnitude")
plt.ylabel("Temperature in log Kelvin")
plt.title("Hertzsprung-Russel Diagram")
plt.scatter(min(log_magnitude), log_temperature[idx_min_mag], marker = "*", color = "black", s = 300, label = "brightest star")
plt.scatter(max(log_magnitude), log_temperature[idx_max_mag], marker = "*", color = "purple", s = 300, label = "dimmest star")
plt.scatter(log_magnitude[idx_max_temp], max(log_temperature), marker = "*", color = "green", s = 300, label = "hottest star")
plt.scatter(log_magnitude[idx_min_temp], min(log_temperature), marker = "*", color = "orange", s = 300, label = "coldest star")
plt.legend()
plt.show()
```
